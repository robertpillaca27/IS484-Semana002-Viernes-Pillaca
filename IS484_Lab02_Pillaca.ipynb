### Representación de un problema mediante estados

## Parte A: Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

### Ficha del Problema: Repartidor de la Ciudad
* **Estado inicial:** El repartidor se encuentra en el Almacén.
* **Estado objetivo:** El repartidor se encuentra en la Zona Norte.
* **Acciones posibles:** Moverse desde la ubicación actual hacia un punto intermedio o zona adyacente que esté conectada directamente.
* **Costo de cada acción:** Distancia (km), tiempo (minutos) o consumo de combustible del trayecto entre dos puntos adyacentes.

### Parte B: Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
"Almacen": ["Centro", "Terminal"],
"Centro": ["Almacen", "ZonaNorte", "Mercado"],
"Terminal": ["Almacen", "Mercado"],
"Mercado": ["Centro", "Terminal", "ZonaNorte"],
"ZonaNorte": ["Centro", "Mercado"],
}

print("Grafo del mapa de la ciudad:")
print(mapa)
```

### Implementación de BFS
### Completar la función de búsqueda en amplitud, usando una cola (lista de Python, con .pop(0) para sacar el primer elemento):

```python
# Capítulo 4.3: Implementación de BFS (Breadth-First Search)
def bfs(grafo, inicio, objetivo): # Define la función BFS con el grafo, nodo inicial y objetivo.
    cola = [[inicio]] # Crea la cola FIFO iniciada con una lista que contiene el nodo inicial.
    visitados = set() # Inicializa un conjunto vacío para almacenar los nodos ya visitados.

    while cola: # Mantiene el bucle activo mientras haya caminos pendientes en la cola.
        camino = cola.pop(0) # Extrae y elimina el primer camino de la cola.
        nodo_actual = camino[-1] # Obtiene el último nodo del camino extraído para evaluarlo.

        if nodo_actual == objetivo: # Comprueba si el nodo actual es el objetivo buscado.
            return camino # Retorna el camino completo encontrado si se alcanzó el objetivo.

        if nodo_actual not in visitados: # Verifica que el nodo actual no se haya procesado antes.
            visitados.add(nodo_actual) # Marca el nodo actual como visitado añadiéndolo al conjunto.
            for vecino in grafo[nodo_actual]: # Iteran los nodos vecinos conectados al nodo actual.
                nuevo_camino = camino + [vecino] # Construye un nuevo camino sumando el vecino al camino actual.
                cola.append(nuevo_camino) # Añade el nuevo camino al final de la cola para explorarlo luego.

    return None # Retorna None si la cola se vacía sin haber encontrado el objetivo.

resultado_bfs = bfs(mapa, "Almacen", "ZonaNorte") # Ejecuta BFS en el 'mapa' desde "Almacen" hasta "ZonaNorte".
print("Camino más corto encontrado con BFS:", resultado_bfs) # Muestra en consola el camino devuelto por la función.
```

### Vectores y matrices con NumPy
### Parte A: Representar cuatro observaciones de un dataset simple (por ejemplo, sensores de temperatura y humedad) como un vector cada una, y juntarlas en una matriz:

```python
import numpy as np # Importa la librería NumPy para manejo de arreglos numéricos.
obs1 = np.array([22.5, 60]) # Crea el primer vector de observación: temperatura y humedad.
obs2 = np.array([19.0, 45]) # Crea el segundo vector de observación.
obs3 = np.array([25.3, 70]) # Crea el tercer vector de observación.
obs4 = np.array([21.0, 50]) # Crea el cuarto vector de observación.
dataset = np.array([obs1, obs2, obs3, obs4]) # Agrupa los 4 vectores en una matriz 2D.
print(dataset) # Imprime la matriz completa en la consola.
print("Forma de la matriz (filas, columnas):", dataset.shape) # Muestra las dimensiones de la matriz.
```

### Parte B: Acceder a una la y a una columna especíca de la matriz:

```python
print("Primera observacion (fila 0):", dataset[0]) # Imprime la primera fila de la matriz (la observación 'obs1').
print("Todas las temperaturas (columna 0):", dataset[:, 0]) # Extrae e imprime los valores de la primera columna (temperaturas).
```

### Producto escalar y distancia euclidiana
### Parte A: Calcular el producto escalar entre dos observaciones:

```python
producto = np.dot(obs1, obs2) # Calcula el producto escalar entre obs1 y obs2.
print("Producto escalar entre obs1 y obs2:", producto) # Imprime el resultado numérico obtenido del producto escalar.
```

### Parte B: Calcular la distancia euclidiana entre dos observaciones, primero con la fórmula manual y luego con NumPy:

```python
# Formula ingresado de forma manual
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2)) # Calcula la distancia euclidiana restando, elevando al cuadrado, sumando y sacando raíz.
print("Distancia manual:", distancia_manual) # Imprime el resultado numérico del cálculo manual de la distancia.
# Con la funcion de NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2) # Calcula la norma del vector diferencia (distancia euclidiana) usando la función de NumPy.
print("Distancia con NumPy:", distancia_numpy) # Imprime el resultado obtenido mediante np.linalg.norm.
```
