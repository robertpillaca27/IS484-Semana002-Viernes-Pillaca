### Representación de un problema mediante estados

## Parte A: Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

### Ficha del Problema: Repartidor de la Ciudad
* **Estado inicial:** El repartidor se encuentra en el Almacén.
* **Estado objetivo:** El repartidor se encuentra en la Zona Norte.
* **Acciones posibles:** Moverse desde la ubicación actual hacia un punto intermedio o zona adyacente que esté conectada directamente.
* **Costo de cada acción:** Distancia (km), tiempo (minutos) o consumo de combustible del trayecto entre dos puntos adyacentes.
