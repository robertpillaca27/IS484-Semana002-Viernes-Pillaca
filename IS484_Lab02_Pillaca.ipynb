* **Estado objetivo:** El repartidor se encuentra en la Zona Norte.
* **Acciones posibles:** Moverse desde la ubicación actual hacia un punto intermedio o zona adyacente que esté conectada directamente.
* **Costo de cada acción:** Distancia (km), tiempo (minutos) o consumo de combustible del trayecto entre dos puntos adyacentes.
