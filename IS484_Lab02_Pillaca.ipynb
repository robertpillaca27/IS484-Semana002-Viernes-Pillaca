** Parte B: Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas
(cada clave es un punto,cada valor es la lista de puntos conectados directamente):**

```python
mapa = {
"Almacen": ["Centro", "Terminal"],
"Centro": ["Almacen", "ZonaNorte", "Mercado"],
"Terminal": ["Almacen", "Mercado"],
"Mercado": ["Centro", "Terminal", "ZonaNorte"],
"ZonaNorte": ["Centro", "Mercado"],
}

print("Grafo del mapa de la ciudad:")
print(mapa)
```

### Implementación de BFS
### Completar la función de búsqueda en amplitud, usando una cola (lista de Python, con .pop(0) para sacar el primer elemento):

```python
# Capítulo 4.3: Implementación de BFS (Breadth-First Search)
def bfs(grafo, inicio, objetivo): # Define la función BFS con el grafo, nodo inicial y objetivo.
    cola = [[inicio]] # Crea la cola FIFO iniciada con una lista que contiene el nodo inicial.
    visitados = set() # Inicializa un conjunto vacío para almacenar los nodos ya visitados.

    while cola: # Mantiene el bucle activo mientras haya caminos pendientes en la cola.
        camino = cola.pop(0) # Extrae y elimina el primer camino de la cola.
        nodo_actual = camino[-1] # Obtiene el último nodo del camino extraído para evaluarlo.

        if nodo_actual == objetivo: # Comprueba si el nodo actual es el objetivo buscado.
            return camino # Retorna el camino completo encontrado si se alcanzó el objetivo.

        if nodo_actual not in visitados: # Verifica que el nodo actual no se haya procesado antes.
            visitados.add(nodo_actual) # Marca el nodo actual como visitado añadiéndolo al conjunto.
            for vecino in grafo[nodo_actual]: # Iteran los nodos vecinos conectados al nodo actual.
                nuevo_camino = camino + [vecino] # Construye un nuevo camino sumando el vecino al camino actual.
                cola.append(nuevo_camino) # Añade el nuevo camino al final de la cola para explorarlo luego.

    return None # Retorna None si la cola se vacía sin haber encontrado el objetivo.

resultado_bfs = bfs(mapa, "Almacen", "ZonaNorte") # Ejecuta BFS en el 'mapa' desde "Almacen" hasta "ZonaNorte".
print("Camino más corto encontrado con BFS:", resultado_bfs) # Muestra en consola el camino devuelto por la función.
```

### Planteamiento del problema
### Se quiere encontrar, dentro de un conjunto de observaciones ya conocidas, cuál es la más parecida a una observación nueva  el mismo principio que usará el algoritmo KNN a partir de la Semana 7.

### Parte A: Implementación de DFS
### Completar la búsqueda en profundidad sobre el mismo mapa de la Sección 4.2, usando una pila(lista de Python, con .pop() para sacar el último elemento):

```python
def dfs(grafo, inicio, objetivo): # Define la función DFS que recibe el grafo, nodo de inicio y objetivo.
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """
    pila = [[inicio]] # Inicializa la pila LIFO con una lista que contiene el nodo inicial.
    visitados = set() # Crea un conjunto vacío para almacenar los nodos ya visitados.
    while pila: # Mantiene el bucle activo mientras existan caminos pendientes en la pila.
        camino = pila.pop() # Saca el último camino agregado a la pila (comportamiento LIFO).
        nodo_actual = camino[-1] # Obtiene el último nodo del camino extraído para analizarlo.
        if nodo_actual == objetivo: # Comprueba si el nodo actual es el objetivo buscado.
            return camino # Retorna el camino encontrado al alcanzar el objetivo.
        if nodo_actual not in visitados: # Verifica que el nodo actual no haya sido procesado antes.
            visitados.add(nodo_actual) # Marca el nodo actual como visitado añadiéndolo al conjunto.
            for vecino in grafo[nodo_actual]: # Recorre cada nodo vecino conectado al nodo actual.
                nuevo_camino = camino + [vecino] # Crea un nuevo camino agregando el vecino al camino previo.
                pila.append(nuevo_camino) # Agrega el nuevo camino al final de la pila (LIFO).
    return None # Retorna None si la pila se vacía sin hallar una ruta al objetivo.

resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte") # Llama a la función DFS para buscar la ruta en 'mapa'.
print("Camino encontrado (DFS):", resultado_dfs) # Imprime en consola el camino devuelto por la función.
```

### Parte B: Encontrar la observación más cercana
### Dado el dataset de la Sección 4.4 (cuatro observaciones de temperatura/humedad) y una observación nueva, calcular la distancia entre la nueva observación y cada una de las existentes, e identicar cuál es la más cercana:

```python
nueva_observacion = np.array([20.5, 48]) # Define la nueva observación de prueba (temperatura y humedad).
distancias = [] # Inicializa una lista vacía para almacenar las distancias calculadas.
for i, obs in enumerate(dataset): # Recorre la matriz 'dataset' obteniendo el índice 'i' y cada fila 'obs'.
    # COMPLETAR: calcular la distancia euclidiana entre 'nueva_observacion' y 'obs'
    d = np.linalg.norm(nueva_observacion - obs) # Calcula la distancia euclidiana entre la nueva observación y la actual.
    distancias.append(d) # Agrega la distancia calculada a la lista de distancias.
    print(f"Distancia a la observacion {i}: {d}") # Muestra en consola el índice de la observación y su distancia.
indice_mas_cercano = np.argmin(distancias) # Encuentra el índice correspondiente al valor mínimo de la lista de distancias.
print("La observacion mas parecida es la numero", indice_mas_cercano) # Imprime el índice de la observación más cercana.
```

### Reflexión Final sobre Distancias y Escalas

1. **¿Por qué la distancia euclidiana es una forma razonable de medir "qué tan parecidas" son dos observaciones?**
   La distancia euclidiana cuantifica la separación geométrica directa entre dos puntos en un espacio multidimensional. Cuando dos observaciones presentan valores numéricos similares en sus características, su posición geométrica está muy cercana, resultando en una distancia pequeña.

2. **¿En qué casos podría no ser la mejor opción?**
   Cuando las características (columnas) del dataset operan en escalas o magnitudes numéricas muy distintas. Por ejemplo, si comparamos *Sueldo mensual en Soles* (rango de 1,000 a 10,000) frente a *Edad en Años* (rango de 18 a 70), las diferencias en Soles inflarán el cálculo numérico y anularán por completo el efecto de la variable Edad.

3. **Solución Técnica:**
   Para solucionar esto, se aplican técnicas de **normalización** (como *MinMaxScaler* para llevar todo al rango [0, 1]) o **estandarización** (como *StandardScaler* para obtener media 0 y varianza 1) antes de calcular la distancia euclidiana.
